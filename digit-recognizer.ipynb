```python
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

import torch
import torch.nn as nn
from torchvision import transforms, models

from torch.autograd import Variable
from torch import nn, optim
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
```

```python
train = pd.read_csv('/kaggle/input/digit-recognizer/train.csv', dtype=np.float32)
test = pd.read_csv('/kaggle/input/digit-recognizer/test.csv', dtype=np.float32)
sample_hub = pd.read_csv('/kaggle/input/digit-recognizer/sample_submission.csv')
```

```python
#pisahkan Target dan Fitur
targets_np = train.label.values
features_np = train.iloc[:, 1:].values/255

#Split data menjadi Train dan Test set

X_train, X_test, y_train, y_test = train_test_split(features_np, targets_np, test_size=0.2, random_state=42)
```

```python
X_train.shape, X_test.shape
```

```python
# Konversi dari Arrays ke Tensors

class DigitDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)
    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```

```python
torch.manual_seed(1111)
train_dataset = DigitDataset(X_train, y_train)
test_dataset = DigitDataset(X_test, y_test)
len(train_dataset)
```

```python
train_dataset.X.shape
```

```python
batch_size = 64
g = torch.Generator()
g.manual_seed(1111)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, generator=g)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=True, generator=g)
```

```python
len(train_loader) #batch size
```

```python
def plot_images(images):
    n_images = len(images)
    rows = int(np.sqrt(n_images))
    cols = int(np.sqrt(n_images))

    fig = plt.figure()
    for i in range(rows*cols):
        ax = fig.add_subplot(rows, cols, i+1)
        ax.imshow(images[i].view(28,28), cmap='bone')
        ax.axis('off')
    #plt.title('handwritten digit')
    plt.show()
```

```python
N_IMAGES = 25

images = [image for image, label in [train_dataset[i] for i in range(N_IMAGES)]]

plot_images(images)
```

```python
class MLPdigit(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 64)
        self.fc5 = nn.Linear(64, output_dim)

        self.dropout = nn.Dropout(p=0.2)
        #self.sftmax = F.softmax

    def forward(self, x):
        x = self.dropout(F.relu(self.fc1(x)))
        x = self.dropout(F.relu(self.fc2(x)))
        x = self.dropout(F.relu(self.fc3(x)))
        x = self.dropout(F.relu(self.fc4(x)))
        x = self.fc5(x)

        return x
```

```python
input_dim = 28 * 28
output_dim = 10
lr=0.001

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model = MLPdigit(input_dim, output_dim)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=lr)

print(f'This model uses {device}')
```

```python
def cal_accuracy(y_pred, y):
    predicted = torch.argmax(y_pred, dim=-1)
    correct = (predicted == y).sum()
    acc = correct.float() / y.shape[0]
    return acc
```

```python
def train(model, train_loader, optimizer, criterion, device):
    epoch_loss = 0.0
    epoch_acc = 0.0

    model.train()

    for X, y in train_loader:
        y_pred = model(X).to(device)        
        
        loss = criterion(y_pred, y)

        acc = cal_accuracy(y_pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()
        epoch_acc += acc.item()

    return epoch_loss/len(train_loader), epoch_acc/len(train_loader)       
```

```python
def evaluate(model, test_loader, criterion, device):
    epoch_loss = 0
    epoch_acc = 0

    model.eval()

    with torch.no_grad():
        for X, y in test_loader:
            y_pred = model(X).to(device)
            loss = criterion(y_pred, y)
            acc = cal_accuracy(y_pred, y)

            epoch_loss += loss.item()
            epoch_acc += acc.item()

    return epoch_loss/len(test_loader), epoch_acc/len(test_loader)
```

```python
best_valid_loss = 0
EPOCHS = 25
train_losses, test_losses, train_accs, valid_accs = [], [], [], []

for epoch in range(EPOCHS):
    train_loss, train_acc = train(model, train_loader, optimizer, criterion, device)
    valid_loss, valid_acc = evaluate(model, test_loader, criterion, device)
    train_losses.append(train_loss)
    test_losses.append(valid_loss)
    train_accs.append(train_acc)
    valid_accs.append(valid_acc)

    if valid_loss < best_valid_loss:
        best_valid_loss = valid_loss
        torch.save(model.state_dict(), 'digit-model.pt')

    print(f'Epoch: {epoch}')
    print(f'\t Train loss: {train_loss:.3f} | Train acc: {train_acc:.3f}')
    print(f'\t  Val  loss: {valid_loss:.3f} |   Val acc: {valid_acc:.3f}')
```

dari hasil eksperimen jumlah epoch ditingkatkan dari 15 hingga 25 epoch untuk mengamati apakah model mengalami overfitting. Evaluasi dilakukan dengan membandingkan train loss & train accuracy terhadap validation loss & validation accuracy pada setiap epoch.

Berdasarkan hasil pelatihan, terlihat bahwa train loss terus menurun secara konsisten dari 0.478 pada epoch ke-0 hingga 0.021 pada epoch ke-24, sementara train accuracy meningkat stabil dari 84.9% menjadi 99.4%. Hal ini menunjukkan bahwa model sangat baik dalam mempelajari data latih.

Namun, pada sisi validasi, validation loss hanya menurun hingga sekitar epoch ke-6–10, dengan nilai terendah berada di kisaran 0.094–0.095, kemudian cenderung meningkat dan berfluktuasi meskipun validation accuracy tetap relatif stabil di kisaran 97%–98%. Kondisi ini menandakan bahwa model mulai menyesuaikan diri secara berlebihan terhadap data latih tanpa peningkatan performa yang signifikan pada data validasi.

Dengan demikian, dapat disimpulkan bahwa indikasi overfitting mulai muncul setelah sekitar epoch ke-10, ditandai dengan:

* Train loss terus menurun
* Validation loss tidak lagi membaik dan cenderung meningkat
* Validation accuracy stagnan

Oleh karena itu, epoch optimal untuk model ini berada di sekitar epoch 6 hingga 10, dan penggunaan early stopping atau regularisasi dapat dipertimbangkan untuk mencegah overfitting pada pelatihan selanjutnya

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# --- Plot 1: Losses ---
ax1.plot(train_losses, label='Training Loss')
ax1.plot(test_losses, label='Validation Loss')
ax1.set_title('Loss vs. Epochs')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True)

# --- Plot 2: Accuracies ---
ax2.plot(train_accs, label='Training Accuracy', color='green')
ax2.plot(valid_accs, label='Validation Accuracy', color='orange')
ax2.set_title('Accuracy vs. Epochs')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()
```
